```python
for col in data.columns:
    print(data[col].unique())
```

```python
data['gender'].replace({'Female':1,'Male':0},inplace=True)
```

```python
for col in data.columns:
    print(data[col].unique())
```

```python
data = pd.get_dummies(data, columns=['country'], dtype=int)
```

```python
for col in data.columns:
    print(f'{col}    {data[col].unique()}')
```

```python
data.dtypes
```

```python
from sklearn.preprocessing import MinMaxScaler
col_to_scale=['estimated_salary','products_number','balance','tenure','age','credit_score']
scaler = MinMaxScaler()
data[col_to_scale] = scaler.fit_transform(data[col_to_scale])
data.sample(5)
```

```python
from sklearn.model_selection import train_test_split
X=data.drop(columns='churn',axis='columns')
y=data['churn']
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=42)
X_train
```

```python
import tensorflow as tf
from tensorflow import keras
model=keras.Sequential([
    keras.layers.Dense(units=8,input_shape=(12,),activation='relu'),
    keras.layers.Dense(units=6,activation='relu'),
    keras.layers.Dense(units=1,activation='sigmoid')

])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.fit(X_train,y_train,epochs=100)
model.evaluate(X_test,y_test)
ypred=model.predict(X_test)
ypred[:10]
```

```python
ypred1 = []

for row in ypred:
    ypred1.append([1 if y >= 0.5 else 0 for y in row])

print(ypred1)
```

```python
from sklearn.metrics import precision_score, recall_score, confusion_matrix



# Calculate precision
precision = precision_score(y_test,ypred1)
print(f"Precision: {precision}")

# Calculate recall
recall = recall_score(y_test, ypred1)
print(f"Recall: {recall}")

# Calculate confusion matrix
conf_matrix = confusion_matrix(y_test, ypred1)
print("Confusion Matrix:")
print(conf_matrix)
```

```python
from sklearn.metrics import f1_score

f1 = f1_score(y_test, ypred1)
print(f"F1 Score: {f1}")
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, ypred1)
print(f"Accuracy: {accuracy}")
tn, fp, fn, tp = conf_matrix.ravel()
specificity = tn / (tn + fp)
print(f"Specificity: {specificity}")
```

```python
from sklearn.metrics import classification_report

print("Classification Report:")
print(classification_report(y_test, ypred1))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, roc_auc_score

# Compute the ROC curve
fpr, tpr, thresholds = roc_curve(y_test, ypred1)
auc_score = roc_auc_score(y_test, ypred1)

plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, label=f"ROC Curve (AUC = {auc_score:.2f})", color='blue')
plt.plot([0, 1], [0, 1], linestyle='--', color='gray')  # Diagonal line
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate (Recall)')
plt.title('ROC Curve')
plt.legend(loc='lower right')
plt.grid()
plt.show()
```

```python
import seaborn as sns
import numpy as np

# Confusion matrix
conf_matrix = confusion_matrix(y_test, ypred1)

plt.figure(figsize=(6, 5))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['Predicted Negative', 'Predicted Positive'],
            yticklabels=['Actual Negative', 'Actual Positive'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```

```python
from sklearn.calibration import calibration_curve

prob_true, prob_pred = calibration_curve(y_test, ypred1, n_bins=10)

plt.figure(figsize=(8, 6))
plt.plot(prob_pred, prob_true, marker='o', color='blue', label='Calibration curve')
plt.plot([0, 1], [0, 1], linestyle='--', color='gray')  # Perfectly calibrated line
plt.xlabel('Predicted Probability')
plt.ylabel('True Probability')
plt.title('Calibration Curve')
plt.grid()
plt.legend()
plt.show()
```
